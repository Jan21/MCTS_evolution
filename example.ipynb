```python
from visualize import Visualizer
import GridEnv
import copy
from partial_plan import PartialPlan

vis = Visualizer()
(grid_env, state) = GridEnv.GridEnv.from_env(0, max_final_component_distance=4)

# Scene 1
state.target = (2, 10)
support = None
pairs = [{"bottleneck": s.bottleneck.position, "support": s.support.position}
         for s, _ in grid_env.propose_subgoal_states(state)]
result = grid_env.solve(state)  
vis.add_scene(
    grid_env,
    state,
    target=state.target,
    target_robot=state.target_robot,
    helpers=state.helpers,
    final_components=list(grid_env.precomputed_final_components[state.target]),
    bottleneck_support_pairs=pairs,
)

# Scene 2
state = copy.deepcopy(state)
state.target = (11, 5)
support = None
pairs = [{"bottleneck": s.bottleneck.position, "support": s.support.position}
         for s, _ in grid_env.propose_subgoal_states(state)]
vis.add_scene(
    grid_env,
    state,
    target=state.target,
    target_robot=state.target_robot,
    helpers=state.helpers,
    final_components=list(grid_env.precomputed_final_components[state.target]),
    bottleneck_support_pairs=pairs,
)

# Scene 3
state = copy.deepcopy(state)
state.target = (11, 9)
state.helpers[0].position = (12, 9)
support = state.helpers[0]
pairs = [{"bottleneck": s.bottleneck.position, "support": s.support.position}
         for s, _ in grid_env.propose_subgoal_states(state, support_robot=support)]
vis.add_scene(
    grid_env,
    state,
    target=state.target,
    target_robot=state.target_robot,
    helpers=state.helpers,
    support=support,
    final_components=list(grid_env._dependent_edge_cache[state.target, support.position]['final_component']),
    bottleneck_support_pairs=pairs,
)

# Scene 4
state = copy.deepcopy(state)
state.target = (10, 10)
state.helpers[0].position = (10, 11)
support = state.helpers[0]
pairs = [{"bottleneck": s.bottleneck.position, "support": s.support.position}
         for s, _ in grid_env.propose_subgoal_states(state, support_robot=support)]
vis.add_scene(
    grid_env,
    target=state.target,
    target_robot=state.target_robot,
    helpers=state.helpers,
    support=support,
    final_components=list(grid_env._dependent_edge_cache[state.target, support.position]['final_component']),
    bottleneck_support_pairs=pairs,
)
result = grid_env.solve(state)

plan = PartialPlan.from_moves(grid_env, state, result["moves"])
plan.save("plan_data.json")  # load in plan_visualize.html
vis.save()
```
